```python
class Employee:
    def addEmpDetails(self,E_ID, E_Name , E_Salary, E_Company):
        self.id=E_ID  #(E_ID is a local variables)
        self.name=E_Name
        self.salary=E_Salary
        self.company=E_Company
        #self.company is an instance variable
    def showEmpDetails(self):
        print("Employee ID:", self.id)
        print("Employee Name:",self.name)
        print("Employee Salary:",self.salary)
        print("Employee Company:",self.company)

e1=Employee()
e1.addEmpDetails(101,"Ansh",1200.00,"HCL")
e1.showEmpDetails()
print("="*25)

e2=Employee()
e2.addEmpDetails(102,"Ansh Sharma",22000.00,"HCL")
e2.showEmpDetails()
print("="*25)

        
```

```python
#OOPS
class Employee:
    def __init__(self,E_ID,E_Name,E_Salary,E_Comapany): #this Keyword
        self.id=E_ID #(E_ID is a local variables)
        self.name=E_Name
        self.salary=E_Salary
        self.company=E_Comapany
        #instance variable

    def showEmpDetails(self):
        print("Employee ID: ",self.id)
        print("Employee Name: ",self.name)
        print("Employee Salary: ",self.salary)
        print("Employee Company: ",self.company)

e1=Employee(101,"Raj",12000.00,"HCL") # __new__()  ___init__()
e1.showEmpDetails()
print("="*25)

e2=Employee(102,"Ansh",15000.00,"HCL")
e2.showEmpDetails()
print("="*25)


# Constructor are used to initializde the value for the current objects.

# Employee e1=new Employee(); in java programming.
      
```
