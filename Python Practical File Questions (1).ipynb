```python
# Q1: ATM Withdrawal Write a Python program for ATM withdrawal.

balance = 5000

pin = int(input("Enter your PIN: "))

if pin == 1234:
    amount = int(input("Enter withdrawal amount: "))

    if amount % 100 == 0:
        if amount <= balance:
            balance = balance - amount
            print("Withdrawal Successful")
            print("Remaining Balance:", balance)
        else:
            print("Insufficient Balance")
    else:
        print("Amount must be a multiple of 100")
else:
    print("Invalid PIN")
```

```python
# Q2: Student Result System 

m1 = int(input("Enter marks of Subject 1: "))
m2 = int(input("Enter marks of Subject 2: "))
m3 = int(input("Enter marks of Subject 3: "))

if m1 >= 40 and m2 >= 40 and m3 >= 40:
    total = m1 + m2 + m3
    percentage = (total / 300) * 100

    print("Percentage:", percentage)

    if percentage >= 75:
        print("Distinction")
    elif percentage >= 60:
        print("First Division")
    elif percentage >= 50:
        print("Second Division")
    else:
        print("Pass")
else:
    print("Fail")
```

```python
# Q10: ATM - Difficult Challenge 


correct_pin = "1234"
balance = 25000
account_active = True

pin = input("Enter your ATM PIN: ")

if pin == correct_pin:
    if account_active:
        amount = int(input("Enter withdrawal amount: "))

        if amount > 0:
            if amount % 100 == 0:
                if amount > 20000:
                    verification = input("Extra verification required. Enter yes/no: ").lower()

                    if verification != "yes":
                        print("Extra Verification Failed")
                        exit()

                if amount <= balance:
                    if balance - amount >= 500:
                        balance = balance - amount
                        print("Withdrawal Successful")
                        print("Withdrawn Amount: ₹", amount)
                        print("Remaining Balance: ₹", balance)
                    else:
                        print("Transaction Failed: Minimum balance of ₹500 must remain")
                else:
                    print("Insufficient Balance")
            else:
                print("Amount must be a multiple of ₹100")
        else:
            print("Amount must be positive")
    else:
        print("Account is inactive")
else:
    print("Incorrect PIN")
 
```

```python
# Q1: ATM Withdrawal - Advanced Challenge 

correct_pin = 1234
pin = int(input("Enter your PIN: "))

balance = 25000
account_active = True

if pin == correct_pin:
    if account_active:
        amount = int(input("Enter withdrawal amount: "))

        if amount > 0:
            if amount % 100 == 0:
                if amount <= balance:
                    if balance - amount >= 500:
                        if amount > 20000:
                            print("Extra verification required")
                            verify = input("Verification successful? (yes/no): ")

                            if verify == "yes":
                                balance = balance - amount
                                print("Withdrawal successful!")
                                print("Remaining balance:", balance)
                            else:
                                print("Verification failed!")
                        else:
                            balance = balance - amount
                            print("Withdrawal successful!")
                            print("Remaining balance:", balance)
                    else:
                        print("Minimum balance of ₹500 must remain")
                else:
                    print("Insufficient balance!")
            else:
                print("Amount must be a multiple of ₹100")
        else:
            print("Withdrawal amount must be positive")
    else:
        print("Account is inactive")
else:
    print("Incorrect PIN")
```

```python
# Q2: Electricity Bill 

units = int(input("Enter electricity units consumed: "))

if units >= 0:
    if units <= 100:
        bill = units * 5
    elif units <= 200:
        bill = units * 7
    elif units <= 300:
        bill = units * 10
    else:
        bill = units * 15

    if bill > 3000:
        surcharge = bill * 0.10
        bill = bill + surcharge

    print("Total Electricity Bill: ₹", bill)

else:
    print("Invalid units!")
```

```python
# Q3: Employee Bonus 

experience = float(input("Enter experience in years: "))
salary = float(input("Enter salary: ₹"))
performance = int(input("Enter performance score: "))

if experience >= 1:
    if salary >= 50000:
        if performance >= 90:
            bonus = salary * 0.20
        elif performance >= 75:
            bonus = salary * 0.10
        else:
            bonus = salary * 0.05
    else:
        if performance >= 90:
            bonus = salary * 0.15
        else:
            bonus = salary * 0.05

    print("Bonus Amount: ₹", bonus)
    print("Total Salary with Bonus: ₹", salary + bonus)

else:
    print("Not eligible for bonus due to insufficient experience.")
```
